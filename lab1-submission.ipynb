5. Create a random numpy array with shape (100, 2) and name it ```z```. Each row indicates the coordinates of a point. Compute point by point distances.

````python
z = np.random.random((100,2))
print(z)
````

````python
for i in range(100):
    for j in range(100):
        if i != j:
            print(np.sqrt(np.sum((z[i] - z[j])**2)))
````
